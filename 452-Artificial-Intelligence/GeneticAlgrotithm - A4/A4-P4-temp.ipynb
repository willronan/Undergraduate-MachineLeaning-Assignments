```python
import numpy as np

def fitness_function(chromosome):
    # insert code
    fitness = 0
    for gene_index in range(len(chromosome)):
        if chromosome[gene_index] == 1:
            fitness = fitness + 1
            
    return fitness

def initialize_population(population_size, chromosome_length):
      # insert code
    
    population = np.zeros((population_size, chromosome_length), dtype=int)
    
    for chromosome in population:
        for gene_index in range(len(chromosome)):
            chromosome[gene_index] = np.random.randint(0, 2)
    
    return population


def roulette_wheel_selection(population, fitness_values):
      # insert code
    total_fitness = np.sum(fitness_values)
    relative_fitness = np.divide(fitness_values, total_fitness)
    
    index = np.random.choice(len(population), p=relative_fitness)
    return population[index]

def single_point_crossover(parent1, parent2, crossover_rate):
      # insert code
    if np.random.choice(2, p=[1-crossover_rate, crossover_rate]) == 1:
        crossover_point = np.random.randint(0,len(parent1))
        child1 = np.concatenate((parent1[:crossover_point], parent2[crossover_point:]))
        child2 = np.concatenate((parent2[:crossover_point], parent1[crossover_point:]))
        return child1, child2
    else:
        return parent1.copy(), parent2.copy()
      
      
def bitwise_mutation(child, mutation_rate):
    for gene_index in range(len(child)):
        if np.random.choice(2, p=[1-mutation_rate, mutation_rate]) == 1:
            child[gene_index] = int(not child[gene_index])

    return child.copy()
        

def genetic_algorithm(population_size, chromosome_length, crossover_rate, mutation_rate, runs):
    generations_to_discovery = []

    
    for run in range(runs):
          # insert code
        population =initialize_population(population_size, chromosome_length)
        fitness = np.zeros(population_size)
        best_fitness = 0
        generation = 0
        
        while best_fitness != 20:

            new_population = np.zeros_like(population, dtype=int)
            generation = generation + 1
        
            for chromosome_index in range(len(population)):
                fitness[chromosome_index] = fitness_function(population[chromosome_index])
            
            best_fitness = np.max(fitness)
            

            for i in range(int(population_size/2)):
                parent1 = roulette_wheel_selection(population, fitness)
                parent2 = roulette_wheel_selection(population, fitness)
                child1, child2 = single_point_crossover(parent1, parent2, crossover_rate)

                new_population[i] = bitwise_mutation(child1, mutation_rate)
                new_population[i+1] = bitwise_mutation(child2, mutation_rate)
                i = i+2

            population = new_population

       
        generations_to_discovery.append(generation)
               

    return np.mean(generations_to_discovery)

# Perform experiments with different mutation and crossover rates
mutation_rates = [0.001, 0.01, 0.1]
crossover_rates = [0.7, 0.5, 0.3]
runs = 20

for mutation_rate in mutation_rates:
    for crossover_rate in crossover_rates:
        avg_generations = genetic_algorithm(100, 20, crossover_rate, mutation_rate, runs)
        print(f"Mutation Rate: {mutation_rate}, Crossover Rate: {crossover_rate}, Average Generations: {avg_generations}")
```
